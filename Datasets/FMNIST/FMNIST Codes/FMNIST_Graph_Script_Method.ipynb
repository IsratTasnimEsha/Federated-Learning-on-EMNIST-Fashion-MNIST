```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

plt.rcParams.update({
    "font.size": 14,
    "axes.labelsize": 16,
    "axes.titlesize": 16,
    "legend.fontsize": 14,
    "xtick.labelsize": 14,
    "ytick.labelsize": 14,
    "lines.linewidth": 2
})

# ---------- Files & Labels ----------
files_csv = [
    "FMNIST_fedprox_federated_metrics_0.csv",
    "FMNIST_masking_federated_metrics_0.csv",
    "FMNIST_masking_federated_metrics_1.csv",
    "FMNIST_masking_federated_metrics_2.csv"
]

labels = [
    "FP",
    "FPM",
    "FPMANR 0.1",
    "FPMANR 0.2"
]

# Professional research color palette
colors = [
    "#1f77b4",  # blue
    "#2ca02c",   # green
    "#ff7f0e",   # orange
    "#d62728"  # red
]

# ---------- Read data ----------
data = []
for f in files_csv:
    df = pd.read_csv(f)
    df = df[df["round"] <= 50]
    data.append(df)

# ---------- Plot helper ----------
def plot_and_save(y_col, y_label, filename, legend_loc, log_scale=False):
    plt.figure(figsize=(7, 5))

    for df, label, color in zip(data, labels, colors):
        plt.plot(
            df["round"],
            df[y_col],
            linestyle='-',      # same solid line for all
            color=color,
            label=label
        )

    if log_scale:
        plt.yscale("log")

    plt.xlabel("Round")
    plt.ylabel(y_label)

    plt.legend(
        loc=legend_loc,
        frameon=True,
        title="Fashion MNIST",
        title_fontsize=14
    )
    plt.setp(plt.gca().get_legend().get_title(), fontweight='bold')

    plt.grid(True, linestyle="--", linewidth=0.5, alpha=0.6)

    plt.tight_layout()
    plt.savefig(filename, dpi=300, bbox_inches="tight")
    plt.close()
    files.download(filename)

# ===============================
#            ACCURACY
# ===============================
plot_and_save("train_acc", "Train Accuracy",
              "comparison_train_accuracy.png", "lower right")

plot_and_save("val_acc", "Validation Accuracy",
              "comparison_validation_accuracy.png", "lower right")

plot_and_save("test_acc", "Test Accuracy",
              "comparison_test_accuracy.png", "lower right")

# ===============================
#              LOSS
# ===============================

# Local client loss
plot_and_save("local_loss", "Local Client Loss",
              "comparison_local_loss.png", "upper right",
              log_scale=True)

# Global train loss
plot_and_save("train_loss", "Global Train Loss",
              "comparison_train_loss.png", "upper right",
              log_scale=True)

# Global validation loss
plot_and_save("val_loss", "Global Validation Loss",
              "comparison_val_loss.png", "upper right",
              log_scale=True)

# Global test loss
plot_and_save("test_loss", "Global Test Loss",
              "comparison_test_loss.png", "upper right",
              log_scale=True)

# ===============================
#              TIME
# ===============================
plot_and_save("round_time", "Round Time (s)",
              "comparison_round_time.png", "lower right")

plot_and_save("total_time", "Total Time (s)",
              "comparison_total_time.png", "lower right")
```
