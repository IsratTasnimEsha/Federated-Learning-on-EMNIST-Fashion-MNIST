```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# ==========================================
# Publication-Quality Plot Settings
# ==========================================
plt.rcParams.update({
    "font.size": 14,
    "axes.labelsize": 16,
    "axes.titlesize": 16,
    "legend.fontsize": 14,
    "xtick.labelsize": 14,
    "ytick.labelsize": 14,
    "lines.linewidth": 2
})

# ==========================================
# Input Files
# ==========================================
file_no_recovery = "EMNIST_Letters_masking_federated_metrics_1_dropout_no_recovery.csv"
file_recovery    = "EMNIST_Letters_masking_federated_metrics_1_dropout_recovery.csv"

# ==========================================
# Load Data
# ==========================================
df_no = pd.read_csv(file_no_recovery)
df_rec = pd.read_csv(file_recovery)

# Use first 50 rounds only
df_no = df_no[df_no["round"] <= 50]
df_rec = df_rec[df_rec["round"] <= 50]

# ==========================================
# Plot: Test Accuracy vs Round
# ==========================================
plt.figure(figsize=(6.5, 4.5))

plt.plot(
    df_no["round"],
    df_no["test_acc"],
    color="#d62728",       # red
    linestyle="-",
    marker=None,
    label="Without Recovery"
)

plt.plot(
    df_rec["round"],
    df_rec["test_acc"],
    color="#1f77b4",       # blue
    linestyle="-",
    marker=None,
    label="With Recovery"
)

plt.xlabel("Round")
plt.ylabel("Test Accuracy (%)")

plt.legend(
    loc="lower right",
    frameon=True,
    title="EMNIST Letters",
    title_fontsize=14
)

plt.setp(
    plt.gca().get_legend().get_title(),
    fontweight="bold"
)

plt.grid(
    True,
    linestyle="--",
    linewidth=0.5,
    alpha=0.6
)

plt.xlim(1, 50)

plt.tight_layout()

# ==========================================
# Save Figure
# ==========================================
filename = "EMNIST_Letters_recovery_vs_no_recovery_test_accuracy.png"

plt.savefig(
    filename,
    dpi=1200,
    bbox_inches="tight"
)

plt.show()

files.download(filename)
```
